# Array Generation

```python
arr=np.arange(1,11,2)
print(arr)
```

```python
arr=np.zeros((6,6))
arr
```

```python
arr=np.ones((6,6))
arr
```

```python
arr=np.linspace(0,1,6)
arr
```

```python
arr=np.linspace(0,1,100)
arr
```

# Randon Array Generation

```python
# normalization : value will remain between 0 - 1
arr=np.random.rand(6)
arr
```

```python
# standardization : value will remain between -3 - 3
arr=np.random.randn(6)
arr
```

```python
arr=np.random.randint(10,20,10)
arr
```
